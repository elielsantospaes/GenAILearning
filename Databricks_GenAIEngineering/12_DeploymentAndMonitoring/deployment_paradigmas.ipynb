# Fundamentals of AI Model Deployment Paradigms

Deployment is the final stage of the MLOps lifecycle, integrating the developed AI model or Compound AI Agent into a production environment where it can generate **completions** (for Generative AI) or **predictions** for downstream users and critical business systems. The choice of deployment paradigm is determined primarily by the required **latency** and **throughput (concurrency)**.

### I. Deployment Paradigms Based on Latency and Throughput

* **Batch Deployment:**
    * **Latency Requirements:** Relatively **low** (i.e., less demanding, allowing for pre-computation of results). Service Level Agreements (SLAs) permit delays ranging from minutes to hours or days.
    * **Throughput/Data Handling:** Suitable for **high throughput** scenarios where large volumes of data can be processed in bulk.
    * **Execution Method:** Jobs are typically **scheduled** to run periodically (e.g., hourly, daily) to compute all predictions ahead of time and **materialize** the results in a storage system.
    * **Generative AI Example:** Summarizing massive corpora of long-form documents (e.g., legal contracts, research papers) to create informational documents used later by a RAG system.
* **Streaming Deployment:**
    * **Latency Requirements:** **Medium** latency requirements, usually within a short time window (e.g., 5 to 15 seconds), faster than batch but not instantaneous.
    * **Throughput/Data Handling:** Handles data in a continuous flow, processing it in **micro-batch** fashion, offering medium throughput.
    * **Execution Method:** Predictions are computed as data is ingested, often triggered by events, and results are materialized or pushed to a real-time data store.
    * **Generative AI Example:** Translating incoming user reviews or comments (e.g., product reviews, social media feeds) as they are published, allowing a short buffer time for the translation to be processed.
* **Real-Time (Online) Deployment:**
    * **Latency Requirements:** **Super low** latency is mandatory, requiring **synchronous** and **instantaneous** responses.
    * **Throughput/Data Handling:** Must handle potentially **high concurrency**, where multiple end-users are requesting completions simultaneously.
    * **Execution Method:** The model or agent is exposed via an API endpoint, and the prediction/completion is generated **on-the-fly** immediately after receiving the user's request.
    * **Generative AI Example:** Interactive applications like **Chatbots**, **AI Assistants**, or **Retrieval-Augmented Generation (RAG) applications** where a human user is waiting for an instant response.

### II. Edge Device Deployment (Specialized Use Case)

* **Definition:** The AI system is deployed locally on the end-user device (e.g., car, smartphone, IoT device) rather than on a remote server.
* **Model Requirements:** This paradigm necessitates highly optimized and constrained models, typically very small **Small Language Models (SLMs)**, often with **less than one or two billion parameters**, to fit within the local memory and processing power constraints.
* **Use Cases:** Suited for simple, low-complexity tasks that require minimal control actions or function offline, such as issuing local commands (e.g., "lower the AC," "open the window"). This is a specialized, hardware-constrained area of ongoing research.

*Note: For production AI Engineering, the primary focus remains on **Batch** and **Real-Time** deployment methodologies, as they cover the vast majority of enterprise Generative AI use cases.*
