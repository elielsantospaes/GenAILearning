# Vector Store
* A vector database is a database that is optimized to store and retrieve your high dimensional vectors, such as embedding.   
* In the RAG architecture contextual information is stored in vectors.   
* Vector database are designed for efficient storage of vectors utilized in generative AI application which rely on identifying documents or image with similarities.
* Vector database provides a query interface that retrieves vector most similar to a specified query vectors.


## Why are Vector Database So Hot?   

* Specialized, full=flegded database for **unstructured data**   
  * Inherit database propertier, i.e: Create-Read-Update-Delete (CRUD)    

* Speed up quey search dor the closest vectors

## How does Vector Search Work?

**Method 1:** Desta Sync API with managed embbeddings
**Feaures:** automatic sync, fully managed embedding   
The embedding are created in a fully managed way. You provide a source delta table and the system calculate the embeddings with one of the fundational embedding models.

**Method 2:** Delta Sync API with self-managed emgbeddings   
**Features:**    
You calculate the embedding with the modelof your choice and provide a source deltatable with the pre calculated embeddings. The system will create the index and the vector DB for you.   

**Method 3:** Direct access CRUD API   
**Features:** upsert operations on the vector database.
