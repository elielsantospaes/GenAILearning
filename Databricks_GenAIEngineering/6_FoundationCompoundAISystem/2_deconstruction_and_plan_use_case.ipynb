# Overview of Modern Generative AI Applications

Today's sophisticated Generative AI applications, such as advanced chatbots or automated research assistants, are built as **Compound AI Systems**. This means the final answer or result is achieved through a coordinated sequence of operations involving the LLM, external tools, and specialized data systems.

The primary motivation for breaking these complex systems down is to:
1.  **Understand Connections:** Clearly map how individual components are linked and interact.
2.  **Define Separation:** Establish precise boundaries and **interfaces** between components, crucial for maintenance and independent development.

### The RAG Architecture Example

The text describes a common and highly effective Compound AI architecture known as **Retrieval-Augmented Generation (RAG)**. RAG breaks the task of answering a question into distinct, sequential stages:

1.  **Question Input:** The process begins with a user's question.
2.  **Search/Retrieval:** The system first calls an external function (like a vector database query) to **retrieve** relevant, up-to-date, or proprietary documents.
3.  **Summary/Compression:** The retrieved documents are often too large to feed directly to the LLM. A function then runs to **summarize** or compress this raw data.
4.  **Context Compilation:** The summaries are assembled into a final, structured **context** which is then included in the prompt.
5.  **Question Answering (QA):** The LLM is called, receiving both the original question and the compiled context, and uses the combined information to generate the **final answer**.
6.  **Answer Output:** The final, informed response is delivered to the user.

## Principles of Component Decomposition

When breaking a compound system into stages, AI engineers must meticulously define four key aspects for every component to ensure a clean and robust architecture:

### 1. Intent (Goal)
* **Definition:** What is the single, clear purpose of this function? (e.g., The intent of the "Search" component is to retrieve external documents relevant to the query.)

### 2. Naming (Conventions)
* **Definition:** How should the component's function be labeled? Following established **naming conventions** (like using the `run_` prefix, e.g., `run_search`) is critical for code readability, consistency, and adherence to software engineering standards, especially in Python.

### 3. Dependencies
* **Definition:** What external libraries, services, or data sources does this component rely on? (e.g., The "Search" component might depend on a **vector database** library.)
* **Importance:** Awareness of dependencies is vital for **deployment** to ensure all necessary libraries are installed in the production environment, preventing runtime failures.

### 4. Signature (Input and Output)
* **Definition:** This defines the precise **input arguments** the function requires and the exact **output** it returns.
* **Role:** The signature acts as the official **interface** between components, ensuring that the output from one stage (e.g., a list of document IDs) matches the required input for the next stage (e.g., the retrieval function).

# Analising each stage in detail.
## Stage 1: The Run Search Component (Retrieval)

The `run_search` stage is the heart of the RAG architecture's "Retrieval" function. Its sole **Intent** is to efficiently and accurately retrieve external, relevant information based on the user's query.

### 1. Component Naming and Intent

* **Intent:** Retrieval of external documents or chunks (vectors) relevant to the user's query.
* **Suggested Naming:** `run_search` (following the established Python convention for operational methods).
* **The System:** This function acts as the bridge between the LLM process and the external data repository.

### 2. Core Dependencies and Tools

The component immediately reveals its need for specific infrastructure:

* **Dependency:** A **Vector Database**. In the example provided, this is assumed to be **Databricks Vector Search**.
* **Required Client:** A client library (e.g., `VectorSearchClient`) is needed to connect to and interact with the database.
* **Key Methods (API Calls):** To execute the retrieval, the component must use specific methods exposed by the vector database client. The necessary steps are:
    1.  **Accessing the Index:** Using a method like `VectorSearchIndex.get_index()` to establish a connection to the specific, indexed collection of documents.
    2.  **Executing Search:** Using a method like `VectorSearchIndex.similarity_search()` to perform the vector comparison and retrieve the closest matches.

### 3. Component Signature (Inputs and Outputs)

Defining the function signature is essential for interface clarity between the RAG stages.

#### Input Arguments:

The function requires the following to successfully execute the search:

* **Query (from User):** The raw text question provided by the end-user. This is the core data used to generate the embedding and perform the similarity search.
* **Endpoint Information:** The component needs credentials or identifiers to connect to the correct vector search instance (e.g., the **Vector Search Endpoint Name** and the **Index Name**).
* **Optional Parameters:** To fine-tune the retrieval process, the function often accepts parameters such as:
    * **Search Column:** Specifying which column within the index to query against.
    * **Filters:** Applying metadata filters (e.g., searching only documents from the last month).
    * **Result Limit:** Configuring the number of documents/chunks to retrieve.

#### Output Result:

* **Data Structure:** The output is typically a **dictionary** or a similar structured object.
* **Content:** This dictionary must contain:
    * **Results:** The retrieved, relevant text chunks or documents.
    * **Pagination Tokens:** Crucial for large result sets, allowing the application to fetch additional results if the initial retrieval limit is too low.

## Stage 2: The Run Summary Component (Compression)

The **Run Summary** stage has the critical responsibility of addressing the fundamental constraint of Large Language Models (LLMs): the **context window limitation**. LLMs can only process a finite amount of input data. If the documents retrieved in the previous stage are too numerous or too long, the system will crash or fail to generate a coherent answer.

### Main Intent and Role

* **Intent:** To compress the raw content retrieved from the Vector Database search into a concise and manageable summary.
* **Role in RAG:** This step serves as the **compression** layer, taking large search results and distilling them into the essential information needed to answer the user's question, ensuring the data fits within the downstream LLM’s context window.

### Technical Dependencies and Implementation

The implementation of the summarization function itself often leverages another specialized LLM, creating a **nested LLM call** within the compound system.

* **Core Dependency:** Since the task is summarization, the component depends on an **LLM service** itself. In the provided example, this involves connecting to the **Databricks Foundational Models API**.
* **Required Client:** A client must be used to communicate with this service (e.g., the `Databricks Deployment Client` or a similar MLflow client configuration).
* **Model Identification:** You must clearly define **which specific LLM model** will be used for the summarization task and its associated **endpoint**. This is a configuration attribute that must be tracked.
* **The Summarization Method:** The method involves feeding the raw content along with a specific **prompt** (e.g., "Summarize the following text") to the summarization LLM. The result is a distilled version of the original document.

### Component Signature (Inputs and Outputs)

Defining the `run_summary` signature formalizes the interface between the retrieval output and the context compilation input.

#### Input Arguments:

* **Content:** The raw text or document(s) retrieved from the preceding `run_search` component.
* **Prompt/Question:** The specific instruction given to the summarization LLM (e.g., "Summarize this content in three key points," or the original user question to bias the summary).
* **LLM Configuration:** Arguments related to the LLM itself, such as temperature, maximum tokens, and model identifiers.

#### Output Result:

* **Summarized Text:** The core output should be the compressed, summarized version of the input content.
* **Consistency:** Importantly, this stage is designed to process **single records** (one document or chunk) at a time, ensuring that the process can be iterated across all retrieved documents.

### Summary of Planning Insight

The key dependency identified in this stage is the need for a **second LLM client and its configuration** to handle the summarization task. This confirms that compound systems require careful management of multiple models, not just the final QA model.

$$\text{Run Summary Component} \implies \text{**Discovered Dependency: Connection and configuration for a Summarization LLM.**}$$

### Stage 3: The Run Augment Component (Parallel Processing)

The **Run Augment Stage** is an optional but highly valuable component focused purely on **system performance and speed**. If the `run_search` stage returns 50 documents, summarizing them sequentially would be slow. This stage solves that problem.

### Main Intent and Role

* **Intent:** To execute the summarization task across **multiple retrieved documents simultaneously (in parallel)**, drastically reducing the total time taken for the system to prepare its context.
* **Role in RAG:** It acts as an orchestrator, managing the flow of data to the `run_summary` component and efficiently handling many summarization tasks at once, ensuring the application remains responsive.

### Technical Dependencies and Implementation

Achieving parallel execution requires specific programming constructs designed for concurrency.

* **Core Concept:** The key idea is **asynchronous processing** or **parallelization**. Instead of waiting for one document to be summarized before starting the next, the system initiates all summarization tasks almost simultaneously.
* **Common Tooling:** In Python, the standard library for managing this kind of concurrency is **`asyncio`**. It allows the system to manage multiple tasks concurrently without being bogged down by waiting for network responses (like the API call to the summarization LLM).
* **Design Note (The Engineering Homework):** Implementing this stage means mastering techniques to:
    1.  Define tasks (one task per document summary).
    2.  Execute tasks concurrently (using features like `asyncio.gather`).
    3.  Track the completion of all parallel tasks.

### Component Signature (Inputs and Outputs)

While the *internal workings* are complex, the inputs and outputs of the `run_augment` component are designed to be simple, acting as a pass-through layer that manages the data from start to finish.

#### Input Arguments:

* **Search Results Dictionary:** The component takes the complete set of results (a dictionary containing multiple documents/records) that were output from the preceding `run_search` component.

#### Output Result:

* **Augmented Dictionary:** The output will also be a dictionary containing all the original and newly created data. Critically, this dictionary now holds the **summarized versions** of all the documents, ready to be passed to the next stage.
* **Interface Goal:** The input and output structures are kept similar to ensure a smooth transition to the upcoming **Run Get Context** component.

### Summary of Planning Insight

The `run_augment` stage introduces the concept of **performance engineering** into the AI system design. It highlights the need to consider not just *what* an LLM can do, but *how fast* it can be done, revealing a key dependency on concurrency libraries.

$$\text{Run Augment Component} \implies \text{**Discovered Dependency: Concurrency frameworks (e.g., Python's asyncio) for speed and scale.**}$$

### Stage 4: The QA Component (Question Answering)

The **QA (Question Answering) component** is where the main Large Language Model (LLM) is finally invoked. Crucially, all previous stages—Search, Summary, and Augment—were essentially sophisticated **data preparation** steps designed to ensure the LLM receives the most relevant and high-quality information possible.

### Main Intent and Role

* **Intent:** To use the compiled context and the original user question to generate the final, authoritative answer.
* **Role in RAG:** This is the **Generation** step of the RAG architecture. It brings all the preparatory work together to produce the system's ultimate output.

### Technical Dependencies and Implementation

Similar to the summarization step, this requires interaction with an LLM service, but with a different focus on the input data.

* **Core Dependency:** The component depends on the **Chat Model Service** (e.g., Databricks Foundational Models API or another endpoint) which runs the final, powerful generative model.
* **Required Client:** The connection is managed via the appropriate client (e.g., the `Databricks Deployment Client`), requiring the correct **endpoint configuration** to be passed.
* **The QA Prompt:** This is a specialized prompt that instructs the LLM on its behavior, typically telling it: **"Based ONLY on the following context, answer the user's question."** This prompt is critical for preventing the model from hallucinating or using its general knowledge base.

### Component Signature (Inputs and Outputs)

The input signature reflects the culmination of the prior stages.

#### Input Arguments:

* **Compiled Context:** The clean, summarized, and relevant data output from the preceding context-building or augment stage.
* **User Question:** The original query entered by the user at the very start of the application flow.
* **QA Prompt:** The system-generated instruction that dictates the LLM's answering behavior.

#### Output Result:

* **Final Answer (Message):** The response generated by the LLM, delivered as a string, which directly addresses the user's question using the provided context.
