```python
from databricks.sdk.service.serving import ChatMessage
from databricks.sdk import WorkspaceClient
from databricks_langchain import ChatDatabricks


#first model for summarization
def query_summary_system(summarization_model,  input:str)->str:
    messages=[
                {
                "role":"system",
                "content":"""You are an assistant that summarizes text.Given a text input,you need to provide a one-sentence summary.You specialize in summiarizing reviews of grocery products. Please keep the reviews in first-person perspective if they're originally written in firs-tperson. Do not change thesentiment.Do not createa                 run-on sentence-be concise."""},
                {
                "role":"user",
                "content":input
                }
                ]
     
    return summarization_model.invoke(messages).content


#secondmodelforsummarization
def challenger_query_summary_system(challenger_model,  input:str)->str:
    messages=[
    {"role":"system",
    "content":"""You are an assistant that summarizes text.Given a text input,you need to provide a one-sentence summary.You specialize in summiarizing reviews of grocery products. Please keep the reviews in first-person perspective if they're originally written in firs-tperson. Do not change thesentiment.Do not createa  run-on sentence-be concise."""
    },
    {
    "role":"user",
    "content":input
    }
    ]
    
    return challenger_model.invoke(messages)
```

```python
summarization_model_endpoint = "databricks-llama-4-maverick"
challenger_model_endpoint = "databricks-gpt-oss-120b"
summarization_model = ChatDatabricks(endpoint = summarization_model_endpoint,
                            temperature = 0,
                            max_tokens = 8096)

challenger_model = ChatDatabricks(endpoint = challenger_model_endpoint,
                            temperature = 0
                            )
```
