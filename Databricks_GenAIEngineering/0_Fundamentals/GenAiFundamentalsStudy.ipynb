# How does LLms Works

### Breaking down LLM in steps   
#### Encoding    
  1 - **Input:**
  The sistem receives data from books, internet like wikpedia, scientific researchs, and custom curated datasets.   

  2 - **Tokenize:**    
  The inputs are converted to numerical representaion   

  3 -**Embedding:**    
  Put the words with similar meaning close in a vector space using a Embedding Function (pre-trained-model)

#### Training    
  The model is trained using a **pre-trained-transformer model**. This steps tipically uses bilions of paratmeters and may envove humam feedback to improve the model in a specific task.

#### Decoding    
  The model convert the output tokens to meaningfull words for us to understand, and predict the next word in a sentence.
